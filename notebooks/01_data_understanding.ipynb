# Prophet

```python
from prophet import Prophet
import matplotlib.pyplot as plt
```

```python
monthly_sales = (
    df.groupby(pd.Grouper(key="Order Date", freq="ME"))["Sales"]
      .sum()
      .reset_index()
)

monthly_sales.head()
```

```python
monthly_sales = monthly_sales.rename(
    columns={
        "Order Date": "ds",
        "Sales": "y"
    }
)

monthly_sales.head()
```

```python
model = Prophet()
```

```python
model.fit(monthly_sales)
```

```python
future = model.make_future_dataframe(
    periods=12,
    freq="ME"
)

future.tail()
```

```python
forecast = model.predict(future)
```

```python
forecast.head()
```

```python
forecast[["ds", "yhat", "yhat_lower", "yhat_upper"]].tail(12)
```

```python
import matplotlib.pyplot as plt

fig = model.plot(forecast)

plt.title("Sales Forecast")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.show()
```

```python
fig2 = model.plot_components(forecast)

plt.show()
```

```python
forecast_result = forecast[
    ["ds", "yhat", "yhat_lower", "yhat_upper"]
]

forecast_result.tail(12)
```

```python
forecast_result.to_csv(
    "../data/processed/sales_forecast.csv",
    index=False
)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

evaluation = monthly_sales.merge(
    forecast[["ds", "yhat"]],
    on="ds"
)

mae = mean_absolute_error(
    evaluation["y"],
    evaluation["yhat"]
)

rmse = np.sqrt(
    mean_squared_error(
        evaluation["y"],
        evaluation["yhat"]
    )
)

print("MAE:", mae)
print("RMSE:", rmse)
```
